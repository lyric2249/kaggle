```python
df3_solve = df3

df3_solve = df3_solve.sort_values(['location', 'date'])

# df3_solve


df3_solve['cum_cases'] = df3_solve.groupby(['location'])['new_cases'].cumsum()


from statsmodels.tsa.stattools import acf
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram

df_wide = df3_solve.pivot(index='date', columns='location', values='new_cases').fillna(0)
# df_wide = df3_solve.pivot(index='date', columns='location', values='cum_cases').fillna(0)

df_wide

# 추천 방식: 리스트에 담아 한 번에 생성
acf_data = []
countries = []

for country in df_wide.columns:
    series = df_wide[country].dropna()
    if len(series) > 50:
        acfs = acf(series, nlags=50)[1:]
        acf_data.append(acfs)
        countries.append(country)

# 리스트를 데이터프레임으로 변환 (컬럼명을 Lag 1~50으로 자동 지정)
acf_matrix = pd.DataFrame(acf_data, index=countries, columns=[f'lag_{i}' for i in range(1, 51)])

dist_matrix = pdist(acf_matrix, metric='euclidean')
linkage_matrix = linkage(dist_matrix, method='ward')


# acf_matrix
```
