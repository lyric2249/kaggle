```python
import inspect
import pkgutil
import importlib
import warnings # 경고 제어 모듈 추가

import sklearn
pkg = sklearn

methods_set = set()
mod_names = [pkg.__name__]

if hasattr(pkg, '__path__'):
    mod_names.extend(name for _, name, _ in pkgutil.walk_packages(pkg.__path__, pkg.__name__ + '.'))

for mod_name in mod_names:
    try:
        # 경고를 무시하는 콘텍스트 매니저 시작
        with warnings.catch_warnings():
            warnings.simplefilter("ignore") # 모든 경고 무시
            mod = importlib.import_module(mod_name)
    except Exception:
        continue
    
    for name, obj in inspect.getmembers(mod):
        if getattr(obj, '__module__', None) != mod_name:
            continue
        
        if inspect.isroutine(obj):
            methods_set.add(f"{mod_name}.{name}")
        elif inspect.isclass(obj):
            for meth_name, meth_obj in inspect.getmembers(obj, inspect.isroutine):
                methods_set.add(f"{mod_name}.{name}.{meth_name}")

import re

for each in set(methods_set):
    if re.search('have', each):
        print(each)
```
