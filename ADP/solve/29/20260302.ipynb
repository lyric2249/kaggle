```python
import pandas as pd
df1 = pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/adp/29/p1.csv', encoding='cp949')


# 1. 정렬 수행 (필수)
df1 = df1.sort_values(by=['계약자고유번호', '거주연도'])

# 2. 그룹별 행 번호 할당
df1['rn'] = df1.groupby('계약자고유번호').cumcount() + 1

df1 = df1[df1['rn']==1]

df1.head(2)

```

```python
import pandas_profiling

profile = df1.profile_report(title='temp', correlations = {'phi_k':False})

profile
```

```python
df1['재계약횟수_bool'] = df1['재계약횟수'].apply(lambda x:'높음' if x>=df1['재계약횟수'].median() else '낮음')

df1.head(2)


```

```python
df4_long = df4

# df4_long = df4_long[]


# ['a1_1', 'a1_2', 'a2_1', 'a2_2', 'a3_1', 'a3_2', 'a4_1',
#        'a4_2', 'a5_1', 'a5_2', 'a6_1', 'a6_2', 'a7_1', 'a7_2', 'a8_1', 'a8_2',
#        'a9_1', 'a9_2', 'b1', 'b2', 'b3', 'b4', 'b5', 'b6', 'b7', 'b8', 'b9']

df4_long = df4_long.melt('game_id')

df4_long['batteridx'] = df4_long['variable'].str.split('_').str[1].fillna('-1').map({'-1':'3_score', '1': '1_first', '2': '2_second'})

df4_long['variable'] = df4_long['variable'].str.split('_').str[0]
df4_long = df4_long[df4_long['value'].isin([1,2,3,4,5,6,7,8,9])]


df4_long['round'] = df4_long['variable'].str.replace('a','').str.replace('b','')



df4_long = df4_long.pivot(index=['game_id', 'round'], columns=['batteridx'], values='value').reset_index()

df4_long.columns.name = None

df4_long = df4_long[(df4_long['1_first']!=4) & (df4_long['2_second']!=4)]

# df4_long.to_csv('temp.csv')

# help(df4_long.reset_index)

# df4_long[['batteridx']]



```
