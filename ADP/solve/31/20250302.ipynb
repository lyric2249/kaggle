```python
df4 =pd.read_csv('https://raw.githubusercontent.com/Datamanim/datarepo/main/adp/31/adp_31_2_v2.csv')
df4.head()
```

```python
df4_solve = df4

df4_solve['bmi'] = df4_solve['weight']/((df4_solve['키']/100)**2)

# df4_solve['생년월일'].to_timestamp()

df4_solve['생년월일'] = pd.to_datetime(df4_solve['생년월일'], format='%Y%m%d')
df4_solve['건강검진일'] = pd.to_datetime(df4_solve['건강검진일'], format='%Y_%m_%d')

df4_solve['age'] = (df4_solve['건강검진일'] - df4_solve['생년월일']).dt.days//365

df4_solve['is_proper'] = (
((df4_solve['성별']=='남성') & (df4_solve['age']==17) & (df4_solve['bmi']>=21.03) & (df4_solve['bmi']<23.21))
|((df4_solve['성별']=='여성') & (df4_solve['age']==17) & (df4_solve['bmi']>=20.03) & (df4_solve['bmi']<22.39))
|((df4_solve['성별']=='남성') & (df4_solve['age']==16) & (df4_solve['bmi']>=21.18) & (df4_solve['bmi']<23.45))
|((df4_solve['성별']=='여성') & (df4_solve['age']==17) & (df4_solve['bmi']>=19.61) & (df4_solve['bmi']<21.74))
)


df4_solve
```
